## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import os
import pandas as pd

possible_paths = [
    'data/small_sample.parquet',
    '../data/small_sample.parquet',
    '../../data/small_sample.parquet',
    '/content/data/small_sample.parquet'
]

df = None
for path in possible_paths:
    if os.path.exists(path):
        df = pd.read_parquet(path)
        print(f"Loaded successfully from: {path}")
        break

if df is not None:
    total_pages = len(df)
    low_ctr_df = df[df['ctr'] < 0.02]
    low_ctr_count = len(low_ctr_df)
    pct_low_ctr = (low_ctr_count / total_pages) * 100

    pos_col = 'avg_position' if 'avg_position' in df.columns else 'position'
    avg_pos_low_ctr = low_ctr_df[pos_col].mean() if pos_col in df.columns else 0

    print("\n--- Real Numbers from Starter Dataset ---")
    print(f"1. Total pages in sample: {total_pages}")
    print(f"2. Low CTR (< 2%) pages: {low_ctr_count} ({pct_low_ctr:.1f}% of total)")
    print(f"3. Average Position for low-CTR pages: {avg_pos_low_ctr:.2f}")
else:
    print("File not found in standard paths. Running fallback with dummy inspection...")
```
